## Parse class and patient (case) ID



```python

def extract_class(path):

    parts = os.path.normpath(path).split(os.sep)
    for part in parts:
        if part.lower() in ["normal", "stroke"]:
            return part.capitalize()
    return None


def extract_case_id(filename):

    match = re.match(r"^\s*(\d+)", filename)
    return match.group(1) if match else None


records = []
for path in image_files:
    filename = os.path.basename(path)
    class_name = extract_class(path)
    case_id = extract_case_id(filename)
    if class_name is None or case_id is None:
        continue
    records.append({
        "image_path": path, "filename": filename,
        "class": class_name, "case_id": case_id,
    })

df = pd.DataFrame(records)
print("Valid image records:", len(df))
print(df.groupby("class")["case_id"].nunique())

```

```python

def calculate_sha256(filepath):
    sha256 = hashlib.sha256()
    with open(filepath, "rb") as f:
        while True:
            chunk = f.read(1024 * 1024)
            if not chunk:
                break
            sha256.update(chunk)
    return sha256.hexdigest()


df["sha256"] = df["image_path"].apply(calculate_sha256)
df["duplicate_of"] = ""
first_occurrence = {}
for index, row in df.iterrows():
    h = row["sha256"]
    if h in first_occurrence:
        df.at[index, "duplicate_of"] = first_occurrence[h]
    else:
        first_occurrence[h] = row["image_path"]

df_unique = df[df["duplicate_of"] == ""].copy()
print("Unique images:", len(df_unique))
print("Unique patients:", df_unique[["class", "case_id"]].drop_duplicates().shape[0])

```

## Preprocessing pipeline

```python
def resize_image(image, size=(224, 224)):
    return cv2.resize(image, size, interpolation=cv2.INTER_AREA)


def apply_clahe(image, clip_limit=2.0, tile_grid_size=(8, 8)):
    gray = image[:, :, 0]
    clahe = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_grid_size)
    enhanced = clahe.apply(gray)
    return np.stack([enhanced] * 3, axis=-1)


def apply_gamma_correction(image, gamma=1.2):
    img_float = image.astype(np.float32) / 255.0
    gamma_corrected = np.power(img_float, gamma)
    return (gamma_corrected * 255.0).astype(np.uint8)


def normalize_image(image):
    return image.astype(np.float32) / 255.0


def preprocess_pipeline(image):
    """Resize -> CLAHE -> gamma correction (gamma=1.2) -> normalize to [0,1]."""
    resized = resize_image(image)
    clahe_img = apply_clahe(resized)
    gamma_img = apply_gamma_correction(clahe_img, gamma=1.2)
    normalized = normalize_image(gamma_img)
    return resized, clahe_img, gamma_img, normalized
```

## Augmentation

```python

train_augmenter = keras.Sequential([
    layers.RandomRotation(factor=8 / 360, fill_mode="constant"),
    layers.RandomZoom(height_factor=(-0.15, 0.15), fill_mode="constant"),
    layers.RandomContrast(factor=0.2),
], name="online_augmentation")


def _load_and_preprocess(path, label):
    def _run(p):
        raw = cv2.imread(p.numpy().decode("utf-8"))
        raw = cv2.cvtColor(raw, cv2.COLOR_BGR2RGB)
        _, _, _, normalized = preprocess_pipeline(raw)
        return normalized.astype(np.float32)

    image = tf.py_function(func=_run, inp=[path], Tout=tf.float32)
    image.set_shape([IMG_SIZE[0], IMG_SIZE[1], 3])
    return image, label


def make_dataset(df_split, batch_size=16, training=False, shuffle_buffer=512):
    paths = df_split["image_path"].values
    labels = (df_split["class"].values == "Stroke").astype(np.float32)

    ds = tf.data.Dataset.from_tensor_slices((paths, labels))
    if training:
        ds = ds.shuffle(shuffle_buffer, seed=RANDOM_SEED, reshuffle_each_iteration=True)
    ds = ds.map(_load_and_preprocess, num_parallel_calls=tf.data.AUTOTUNE)
    if training:
        ds = ds.map(lambda x, y: (train_augmenter(x, training=True), y),
                    num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.batch(batch_size)
    ds = ds.prefetch(tf.data.AUTOTUNE)
    return ds

```

## Vision-Mamba model

```python
class SelectiveSSSP(layers.Layer):

    def __init__(self, dim, state_dim=16, **kwargs):
        super().__init__(**kwargs)

        self.dim = dim
        self.state_dim = state_dim

        self.parameter_projection = layers.Dense(
            self.state_dim * 2 + self.dim,
            name="parameter_projection"
        )

        self.delta_projection = layers.Dense(
            self.dim,
            name="delta_projection"
        )

        self.A_log = self.add_weight(
            name="A_log",
            shape=(self.dim, self.state_dim),
            initializer="ones",
            trainable=True
        )

        self.D = self.add_weight(
            name="D",
            shape=(self.dim,),
            initializer="ones",
            trainable=True
        )

    def call(self, x):

        # ------------------------------------------------
        # Parameter Generation
        # Generates B_t, C_t and Delta_t
        # ------------------------------------------------

        proj = self.parameter_projection(x)

        B_t, C_t, delta_pre = tf.split(
            proj,
            [
                self.state_dim,
                self.state_dim,
                self.dim
            ],
            axis=-1
        )

        # ------------------------------------------------
        # Input-dependent discretization step
        # Delta_t = softplus(...)
        # ------------------------------------------------

        delta = tf.nn.softplus(
            self.delta_projection(delta_pre)
        )

        # ------------------------------------------------
        # Learnable continuous state-transition parameter
        # A
        # ------------------------------------------------

        A = -tf.exp(self.A_log)

        # ------------------------------------------------
        # Discretization
        # A_bar = exp(Delta * A)
        # B_bar = Delta * B
        # ------------------------------------------------

        delta_exp = tf.expand_dims(
            delta,
            axis=-1
        )

        A_bar = tf.exp(
            delta_exp * A
        )

        B_bar = (
            delta_exp *
            tf.expand_dims(B_t, axis=2)
        )

        # ------------------------------------------------
        # Initial hidden state
        # ------------------------------------------------

        batch = tf.shape(x)[0]

        h0 = tf.zeros(
            (
                batch,
                self.dim,
                self.state_dim
            ),
            dtype=x.dtype
        )

        # ------------------------------------------------
        # Sequential State Update
        #
        # h_t = A_bar_t h_(t-1)
        #       + B_bar_t x_t
        # ------------------------------------------------

        x_seq = tf.transpose(
            x,
            [1, 0, 2]
        )

        A_bar_seq = tf.transpose(
            A_bar,
            [1, 0, 2, 3]
        )

        B_bar_seq = tf.transpose(
            B_bar,
            [1, 0, 2, 3]
        )

        def scan_fn(h_prev, elems):

            A_t, B_t_step, x_t = elems

            return (
                A_t * h_prev
                +
                B_t_step *
                tf.expand_dims(x_t, axis=-1)
            )

        h_seq = tf.scan(
            scan_fn,
            (
                A_bar_seq,
                B_bar_seq,
                x_seq
            ),
            initializer=h0
        )

        h_seq = tf.transpose(
            h_seq,
            [1, 0, 2, 3]
        )

        # ------------------------------------------------
        # Output Generation
        #
        # y_t = C_t h_t + D x_t
        # ------------------------------------------------

        y = tf.reduce_sum(
            h_seq *
            tf.expand_dims(C_t, axis=2),
            axis=-1
        )

        y = y + self.D * x

        return y
```

```python
class SelectiveSSMBlock(layers.Layer):

    def __init__(self, dim, state_dim=16, **kwargs):
        super().__init__(**kwargs)

        self.input_norm = layers.LayerNormalization(
            name="input_layer_norm"
        )

        self.input_projection = layers.Dense(
            dim,
            name="input_projection"
        )

        self.sssp = SelectiveSSSP(
            dim=dim,
            state_dim=state_dim,
            name="selective_sssp"
        )

        self.output_projection = layers.Dense(
            dim,
            name="output_projection"
        )

        self.gate_projection = layers.Dense(
            dim,
            name="gate_projection"
        )

        self.output_norm = layers.LayerNormalization(
            name="output_layer_norm"
        )

    def call(self, x):

        residual = x

        x = self.input_norm(x)

        x_in = self.input_projection(x)

        x_in = tf.nn.gelu(x_in)

        y = self.sssp(x_in)

        y = self.output_projection(y)

        z = tf.nn.sigmoid(
            self.gate_projection(y)
        )

        y_gated = y * z

        output = residual + y_gated

        output = self.output_norm(output)

        return output
```

```python
def build_vision_mamba_branch(
    input_shape=(224, 224, 3),
    patch_size=4,
    embed_dim=384,
    n_blocks=24,
    state_dim=16,
    target_spatial=14
):

    inputs = layers.Input(shape=input_shape)

    tokens = layers.Conv2D(
        embed_dim,
        kernel_size=patch_size,
        strides=patch_size,
        padding="valid",
        name="mamba_patch_embed"
    )(inputs)

    grid_h = tokens.shape[1]
    grid_w = tokens.shape[2]

    x = layers.Reshape(
        (grid_h * grid_w, embed_dim),
        name="flatten_sequence"
    )(tokens)

    for i in range(n_blocks):

        x = SelectiveSSMBlock(
            embed_dim,
            state_dim=state_dim,
            name=f"ssm_block_{i}"
        )(x)

    x = layers.LayerNormalization(
        name="mamba_out_norm"
    )(x)

    x = layers.Reshape(
        (grid_h, grid_w, embed_dim),
        name="mamba_feature_map"
    )(x)

    x = layers.AveragePooling2D(
        pool_size=(
            grid_h // target_spatial,
            grid_w // target_spatial
        ),
        name="mamba_adaptive_avg_pool"
    )(x)

    return keras.Model(
        inputs,
        x,
        name="vision_mamba_branch"
    )
```

## AHDA module (Adaptive Hemispheric Difference Attention)

```python
import tensorflow as tf
from tensorflow.keras import layers


class AHDA(layers.Layer):


    def __init__(self, channels=384, reduction=8, **kwargs):
        super().__init__(**kwargs)

        self.channels = channels
        self.reduction = reduction

        # FEATURE ALIGNMENT


        self.align_conv = layers.Conv2D(
            filters=channels,
            kernel_size=1,
            strides=1,
            padding="same",
            name="ahda_align_conv"
        )


        # ADAPTIVE ATTENTION GENERATION


        hidden_dim = max(
            channels // reduction,
            8
        )

        self.fc1 = layers.Dense(
            hidden_dim,
            activation="gelu",
            name="ahda_fc1"
        )

        self.fc2 = layers.Dense(
            channels,
            activation="sigmoid",
            name="ahda_fc2"
        )


        # Global Average Pooling


        self.gap = layers.GlobalAveragePooling2D(
            name="ahda_gap"
        )


        # Final Layer Normalization


        self.output_norm = layers.LayerNormalization(
            name="ahda_output_norm"
        )

    def call(self, f_local, f_context):


        concatenated = tf.concat(
            [
                f_local,
                f_context
            ],
            axis=-1
        )


        aligned = self.align_conv(
            concatenated
        )

        # GELU activation
        aligned = tf.nn.gelu(
            aligned
        )

        # Save aligned feature for residual learning
        f_aligned = aligned



        width = tf.shape(
            aligned
        )[2]

        midline = width // 2

        f_left = aligned[
            :,
            :,
            :midline,
            :
        ]

        f_right = aligned[
            :,
            :,
            midline:,
            :
        ]



        f_right_ref = tf.reverse(
            f_right,
            axis=[2]
        )



        hemispheric_difference = tf.abs(
            f_left - f_right_ref
        )




        z = self.gap(
            hemispheric_difference
        )



        z = self.fc1(z)


        attention = self.fc2(z)



        attention = tf.reshape(
            attention,
            (
                -1,
                1,
                1,
                self.channels
            )
        )



        recalibrated = (
            f_aligned * attention
        )



        enhanced = (
            recalibrated
            +
            f_aligned
        )


        output = self.output_norm(
            enhanced
        )

        return output
```

## ACAF module (Adaptive Cross-Attention Fusion)

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers


class ACAF(layers.Layer):

    def __init__(
        self,
        channels=384,
        ffn_dim=None,
        **kwargs
    ):
        super().__init__(**kwargs)

        self.channels = channels

        if ffn_dim is None:
            ffn_dim = channels * 4

        self.ffn_dim = ffn_dim

        self.proj_local = layers.Dense(
            channels,
            name="acaf_proj_local"
        )

        self.proj_ahda = layers.Dense(
            channels,
            name="acaf_proj_ahda"
        )

        self.query_projection = layers.Dense(
            channels,
            name="query_projection"
        )

        self.key_projection = layers.Dense(
            channels,
            name="key_projection"
        )

        self.value_projection = layers.Dense(
            channels,
            name="value_projection"
        )

        self.norm1 = layers.LayerNormalization(
            name="acaf_attention_norm"
        )

        self.ffn_dense1 = layers.Dense(
            ffn_dim,
            activation="gelu",
            name="acaf_ffn_dense1"
        )

        self.ffn_dense2 = layers.Dense(
            channels,
            name="acaf_ffn_dense2"
        )

        self.norm2 = layers.LayerNormalization(
            name="acaf_ffn_norm"
        )

    def call(self, f_local, f_ahda):

        batch_size = tf.shape(f_local)[0]

        height = tf.shape(f_local)[1]
        width = tf.shape(f_local)[2]

        f_local_seq = tf.reshape(
            f_local,
            (
                batch_size,
                height * width,
                self.channels
            )
        )

        f_ahda_seq = tf.reshape(
            f_ahda,
            (
                batch_size,
                height * width,
                self.channels
            )
        )

        projected_local = self.proj_local(
            f_local_seq
        )

        projected_ahda = self.proj_ahda(
            f_ahda_seq
        )

        Q = self.query_projection(
            projected_local
        )

        K = self.key_projection(
            projected_ahda
        )

        V = self.value_projection(
            projected_ahda
        )

        scale = tf.sqrt(
            tf.cast(
                self.channels,
                Q.dtype
            )
        )

        attention_scores = tf.matmul(
            Q,
            K,
            transpose_b=True
        )

        attention_scores = (
            attention_scores / scale
        )

        attention_weights = tf.nn.softmax(
            attention_scores,
            axis=-1
        )

        cross_attended = tf.matmul(
            attention_weights,
            V
        )

        refined = self.norm1(
            Q + cross_attended
        )

        ffn_output = self.ffn_dense1(
            refined
        )

        ffn_output = self.ffn_dense2(
            ffn_output
        )

        fused = self.norm2(
            refined + ffn_output
        )

        output = tf.reshape(
            fused,
            (
                batch_size,
                height,
                width,
                self.channels
            )
        )

        return output
```

```python
def get_convnext_stage3_output(convnext_model, target_shape=(14, 14, 384)):
    for layer in reversed(convnext_model.layers):
        try:
            shape = tuple(layer.output.shape[1:])   # was layer.output_shape
        except Exception:
            continue
        if shape == target_shape:
            return layer.output
    raise ValueError(f"No layer with output shape {target_shape}")


def list_convnext_layer_shapes(convnext_model):

    for layer in convnext_model.layers:
        try:
            shape = layer.output_shape
        except Exception as e:
            print(f"{layer.name:40s}  <no single output_shape: {e}>")
            continue
        if shape and len(shape) == 4:
            print(f"{layer.name:40s}  {shape}")
```

## Full proposed model: ConvNeXt-Tiny + Vision Mamba + AHDA + ACAF

```python

def build_proposed_model(dropout_rate=0.30, mamba_embed_dim=384, mamba_blocks=24,
                          mamba_state_dim=16, input_shape=(224, 224, 3),
                          backbone_trainable=True):
    inputs = layers.Input(shape=input_shape, name="ct_image")

    # ConvNeXt-Tiny local branch
    convnext = keras.applications.ConvNeXtTiny(include_top=False, weights="imagenet", input_shape=input_shape)
    convnext.trainable = backbone_trainable
    stage3_output = get_convnext_stage3_output(convnext, target_shape=(14, 14, 384))
    convnext_stage3 = keras.Model(convnext.input, stage3_output, name="convnext_tiny_stage3")
    f_local = convnext_stage3(layers.Rescaling(255.0, name="to_0_255")(inputs))

    # Vision-Mamba
    mamba_branch = build_vision_mamba_branch(
        input_shape=input_shape, embed_dim=mamba_embed_dim,
        n_blocks=mamba_blocks, state_dim=mamba_state_dim, target_spatial=14,
    )
    f_context = mamba_branch(inputs)

    # AHDA
    f_ahda = AHDA(channels=mamba_embed_dim, name="ahda_module")(f_local, f_context)

    # ACAF
    f_fused = ACAF(channels=mamba_embed_dim, name="acaf_module")(f_local, f_ahda)

    # Classification head
    pooled = layers.GlobalAveragePooling2D(name="gap")(f_fused)
    pooled = layers.Dropout(dropout_rate, name="dropout")(pooled)
    outputs = layers.Dense(1, activation="sigmoid", name="stroke_probability")(pooled)
    model = keras.Model(inputs, outputs, name="Proposed_ConvNeXt_VisionMamba_AHDA_ACAF")
    return model, convnext

```

```python
import os, time, json, platform, warnings, logging
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from scipy import stats
from scipy.stats import norm
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import log_loss
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, ConstantKernel, WhiteKernel

warnings.filterwarnings("ignore")
tf.get_logger().setLevel(logging.ERROR)

# CONFIG
FITNESS_MODE = "single_split"
K_TUNE = 5
N_EVALUATIONS = 96
N_INITIAL = 10
N_EI_CANDIDATES = 5000
EI_XI = 0.01
EPOCHS_PER_CANDIDATE = 50
PATIENCE = 8
SEEDS = [42, 43, 44, 45, 46]
RESUME = False
METHOD_NAME = "Bayesian Optimization"
LOG_PREFIX = "bayes_opt"

SEARCH_SPACE = {
    "learning_rate": (1e-5, 1e-3),
    "batch_size": [8, 16, 32],
    "weight_decay": (1e-5, 1e-3),
    "dropout_rate": (0.10, 0.50),
}


.
BUILD_FN = build_proposed_model


def compile_model(model, learning_rate, weight_decay):
    optimizer = keras.optimizers.AdamW(learning_rate=learning_rate, weight_decay=weight_decay)
    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=[keras.metrics.BinaryAccuracy(name="accuracy")],
    )


# Folds
if FITNESS_MODE == "cv_patient":
    assert train_val_df.groupby("case_id")["class"].nunique().max() == 1, "case_id maps to >1 class!"
    y_tv = (train_val_df["class"] == "Stroke").astype(int).values
    skf = StratifiedGroupKFold(n_splits=K_TUNE, shuffle=True, random_state=42)
    tune_folds = list(skf.split(train_val_df, y_tv, train_val_df["case_id"]))
    for i, (tr, va) in enumerate(tune_folds, 1):
        assert set(train_val_df["case_id"].iloc[tr]).isdisjoint(set(train_val_df["case_id"].iloc[va]))


# Fitness functions
def single_split_fitness(learning_rate, batch_size, weight_decay, dropout_rate):
    """Same fitness as the GTO / GTO+Levy / ALGTO searches: best val_loss on val_df."""
    keras.backend.clear_session()
    train_ds = make_dataset(train_df, batch_size=batch_size, training=True)
    val_ds = make_dataset(val_df, batch_size=batch_size, training=False)

    model, _ = BUILD_FN(dropout_rate=dropout_rate)
    compile_model(model, learning_rate, weight_decay)
    history = model.fit(
        train_ds, validation_data=val_ds, epochs=EPOCHS_PER_CANDIDATE, verbose=0,
        callbacks=[keras.callbacks.EarlyStopping(monitor="val_loss", patience=PATIENCE,
                                                 restore_best_weights=True)],
    )
    best = float(np.min(history.history["val_loss"]))
    return best, [best]


def cv_fitness(learning_rate, batch_size, weight_decay, dropout_rate):
    """Mean over K folds of the PATIENT-level log loss (mean image probability per patient)."""
    fold_losses = []
    for tr_idx, va_idx in tune_folds:
        keras.backend.clear_session()
        fold_train = train_val_df.iloc[tr_idx].reset_index(drop=True)
        fold_val = train_val_df.iloc[va_idx].reset_index(drop=True)

        train_ds = make_dataset(fold_train, batch_size=batch_size, training=True)
        val_ds = make_dataset(fold_val, batch_size=batch_size, training=False)

        model, _ = BUILD_FN(dropout_rate=dropout_rate)
        compile_model(model, learning_rate, weight_decay)
        model.fit(
            train_ds, validation_data=val_ds, epochs=EPOCHS_PER_CANDIDATE, verbose=0,
            callbacks=[keras.callbacks.EarlyStopping(monitor="val_loss", patience=PATIENCE,
                                                     restore_best_weights=True)],
        )

        img_prob = model.predict(val_ds, verbose=0).ravel()       # val_ds is not shuffled
        tmp = pd.DataFrame({
            "case_id": fold_val["case_id"].values,
            "y_true": (fold_val["class"] == "Stroke").astype(int).values,
            "y_prob": img_prob,
        })
        pat = tmp.groupby("case_id").agg(y_true=("y_true", "first"),
                                         y_prob=("y_prob", "mean")).reset_index()
        p = np.clip(pat["y_prob"].values, 1e-7, 1 - 1e-7)
        fold_losses.append(float(log_loss(pat["y_true"].values, p, labels=[0, 1])))

    return float(np.mean(fold_losses)), fold_losses


fitness_fn = single_split_fitness if FITNESS_MODE == "single_split" else cv_fitness


# Bayesian optimization
LR_LOG = np.log10(SEARCH_SPACE["learning_rate"])
WD_LOG = np.log10(SEARCH_SPACE["weight_decay"])
DO_LO, DO_HI = SEARCH_SPACE["dropout_rate"]
BATCHES = SEARCH_SPACE["batch_size"]


def decode(u):
    """Unit cube [0,1]^4 -> hyperparameters (learning rate and weight decay on log scale)."""
    u = np.clip(u, 0.0, 1.0)
    return {
        "learning_rate": float(10 ** (LR_LOG[0] + u[0] * (LR_LOG[1] - LR_LOG[0]))),
        "batch_size": int(BATCHES[min(int(u[1] * len(BATCHES)), len(BATCHES) - 1)]),
        "weight_decay": float(10 ** (WD_LOG[0] + u[2] * (WD_LOG[1] - WD_LOG[0]))),
        "dropout_rate": float(DO_LO + u[3] * (DO_HI - DO_LO)),
    }


def encode(p):
    """Hyperparameters -> unit cube (used to rebuild the GP training data from the log)."""
    return np.array([
        (np.log10(p["learning_rate"]) - LR_LOG[0]) / (LR_LOG[1] - LR_LOG[0]),
        (BATCHES.index(int(p["batch_size"])) + 0.5) / len(BATCHES),
        (np.log10(p["weight_decay"]) - WD_LOG[0]) / (WD_LOG[1] - WD_LOG[0]),
        (p["dropout_rate"] - DO_LO) / (DO_HI - DO_LO),
    ])


def expected_improvement(mu, sigma, best, xi=EI_XI):
    sigma = np.maximum(sigma, 1e-9)
    imp = best - mu - xi                       # we MINIMIZE the loss
    z = imp / sigma
    return imp * norm.cdf(z) + sigma * norm.pdf(z)


def suggest_next(X, y, rng, gp_seed):
    kernel = (ConstantKernel(1.0, (1e-3, 1e3)) * Matern(length_scale=np.ones(4), nu=2.5,
              length_scale_bounds=(1e-2, 1e2)) + WhiteKernel(1e-4, (1e-8, 1e-1)))
    gp = GaussianProcessRegressor(kernel=kernel, normalize_y=True,
                                  n_restarts_optimizer=3, random_state=gp_seed)
    gp.fit(np.asarray(X), np.asarray(y))
    cand = rng.random((N_EI_CANDIDATES, 4))
    mu, sd = gp.predict(cand, return_std=True)
    ei = expected_improvement(mu, sd, best=float(np.min(y)))
    return cand[int(np.argmax(ei))]


def run_bayes_opt(seed, log_path):
    if os.path.exists(log_path) and not RESUME:
        os.remove(log_path)                      # never reuse a stale/debug log
    records = pd.read_csv(log_path).to_dict("records") if os.path.exists(log_path) else []
    init_points = np.random.default_rng(seed).random((N_INITIAL, 4))     # deterministic per seed

    for i in range(len(records), N_EVALUATIONS):
        if i < N_INITIAL:
            u, phase = init_points[i], "initial_random"
        else:
            X = [encode(r) for r in records]
            y = [r["cv_val_loss"] for r in records]
            u = suggest_next(X, y, np.random.default_rng(seed * 1000 + i), gp_seed=seed * 1000 + i)
            phase = "bayesian"
        params = decode(u)

        tf.keras.utils.set_random_seed(seed * 1000 + i)
        t0 = time.time()
        fit, fold_losses = fitness_fn(**params)
        prev_best = min([r["cv_val_loss"] for r in records], default=np.inf)
        records.append({
            "seed": seed, "evaluation": i + 1, "phase": phase, **params,
            "cv_val_loss": fit,
            **{f"fold{k+1}_val_loss": l for k, l in enumerate(fold_losses)},
            "best_so_far": min(fit, prev_best),
            "seconds": round(time.time() - t0, 1),
        })
        pd.DataFrame(records).to_csv(log_path, index=False)      # save after every trial

    log = pd.DataFrame(records)
    best = log.loc[log["cv_val_loss"].idxmin()]
    return best, log



best_by_seed, per_seed = {}, []
for s in SEEDS:
    best, log = run_bayes_opt(s, f"{LOG_PREFIX}_{FITNESS_MODE}_seed{s}_log.csv")
    best_by_seed[s] = best
    fold_cols = [c for c in log.columns if c.startswith("fold") and c.endswith("_val_loss")]
    per_seed.append({
        "seed": s,
        "best_val_loss": float(best["cv_val_loss"]),
        "fold_losses": [float(best[c]) for c in fold_cols],
        "time_min": float(log["seconds"].sum() / 60.0),
    })



COLS = ["Optimization Strategy", "Best Validation Loss (Mean ± SD)",
        "95% CI", "Mean Optimization Time per Run (min)"]


def ci95(values):
    v = np.asarray(values, float)
    n = len(v)
    if n < 2:
        return float(v.mean()), np.nan, np.nan, np.nan
    mean, sd = v.mean(), v.std(ddof=1)
    half = stats.t.ppf(0.975, n - 1) * sd / np.sqrt(n)
    return mean, sd, mean - half, mean + half


def fmt_ms(m, sd):
    return f"{m:.4f}" if np.isnan(sd) else f"{m:.4f} ± {sd:.4f}"


def fmt_ci(lo, hi):
    return "–" if np.isnan(lo) else f"{lo:.4f}–{hi:.4f}"



seed_rows = []
for r in per_seed:
    m, sd, lo, hi = ci95(r["fold_losses"])
    seed_rows.append({
        COLS[0]: f"{METHOD_NAME} (seed {r['seed']})",
        COLS[1]: fmt_ms(m, sd),
        COLS[2]: fmt_ci(lo, hi),
        COLS[3]: f"{r['time_min']:,.0f}",
    })
seed_table = pd.DataFrame(seed_rows, columns=COLS)


vals = [r["best_val_loss"] for r in per_seed]
times = [r["time_min"] for r in per_seed]
m, sd, lo, hi = ci95(vals)
overall_table = pd.DataFrame([{
    COLS[0]: METHOD_NAME,
    COLS[1]: fmt_ms(m, sd),
    COLS[2]: fmt_ci(lo, hi),
    COLS[3]: f"{np.mean(times):,.0f}",
}], columns=COLS)


print("\t".join(COLS[1:]))
print("\t".join(str(overall_table[c].iloc[0]) for c in COLS[1:]))

seed_table.to_csv(f"{LOG_PREFIX}_per_seed_table.csv", index=False)
overall_table.to_csv(f"table12_{LOG_PREFIX}.csv", index=False)
pd.DataFrame([{k: v for k, v in r.items() if k != "fold_losses"} for r in per_seed]) \
  .to_csv(f"{LOG_PREFIX}_per_seed_values.csv", index=False)



overall_best = min(best_by_seed.values(), key=lambda b: b["cv_val_loss"])
bo_best_params = {
    "learning_rate": float(overall_best["learning_rate"]),
    "batch_size": int(overall_best["batch_size"]),
    "weight_decay": float(overall_best["weight_decay"]),
    "dropout_rate": float(overall_best["dropout_rate"]),
}

json.dump({"python": platform.python_version(), "tensorflow": tf.__version__,
           "gpus": [g.name for g in tf.config.list_physical_devices("GPU")],
           "FITNESS_MODE": FITNESS_MODE, "K_TUNE": K_TUNE, "N_EVALUATIONS": N_EVALUATIONS,
           "N_INITIAL": N_INITIAL, "N_EI_CANDIDATES": N_EI_CANDIDATES, "EI_XI": EI_XI,
           "surrogate": "GaussianProcessRegressor (Matern 2.5 ARD + White noise, normalize_y)",
           "acquisition": "Expected Improvement (minimization)",
           "EPOCHS_PER_CANDIDATE": EPOCHS_PER_CANDIDATE, "PATIENCE": PATIENCE,
           "SEEDS": SEEDS, "SEARCH_SPACE": SEARCH_SPACE},
          open(f"{LOG_PREFIX}_environment.json", "w"), indent=2)
json.dump(bo_best_params, open(f"{LOG_PREFIX}_best_config.json", "w"), indent=2)
```

```python
# ============================================================
# BAYESIAN OPTIMIZATION


import os
import time
import json
import platform
import warnings
import logging

import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from scipy import stats

from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import log_loss

from skopt import Optimizer
from skopt.space import Real, Integer, Categorical


warnings.filterwarnings("ignore")
tf.get_logger().setLevel(logging.ERROR)


# CONFIGURATION


SEEDS = [42, 43, 44, 45, 46]

K_TUNE = 5

N_EVALUATIONS = 96

EPOCHS_PER_CANDIDATE = 50

PATIENCE = 8

RESUME = False

METHOD_NAME = "Bayesian Optimization"

LOG_PREFIX = "bayesian_optimization_patient"


BUILD_FN = build_proposed_model




train_val_df = pd.concat(
    [train_df, val_df],
    ignore_index=True
)




assert (
    train_val_df
    .groupby("case_id")["class"]
    .nunique()
    .max()
    == 1
), "A case_id contains more than one class!"



def compile_model(
    model,
    learning_rate,
    weight_decay
):

    optimizer = keras.optimizers.AdamW(
        learning_rate=learning_rate,
        weight_decay=weight_decay
    )

    model.compile(
        optimizer=optimizer,
        loss="binary_crossentropy",
        metrics=[
            keras.metrics.BinaryAccuracy(
                name="accuracy"
            ),
            keras.metrics.Precision(
                name="precision"
            ),
            keras.metrics.Recall(
                name="recall"
            ),
            keras.metrics.AUC(
                name="auc"
            ),
        ],
    )



SEARCH_SPACE = [
    Real(
        1e-5,
        1e-3,
        prior="log-uniform",
        name="learning_rate"
    ),

    Categorical(
        [8, 16, 32],
        name="batch_size"
    ),

    Real(
        1e-5,
        1e-3,
        prior="log-uniform",
        name="weight_decay"
    ),

    Real(
        0.10,
        0.50,
        prior="uniform",
        name="dropout_rate"
    ),
]




y_tv = (
    train_val_df["class"] == "Stroke"
).astype(int).values

groups_tv = train_val_df["case_id"].values


def create_tuning_folds(seed):

    skf = StratifiedGroupKFold(
        n_splits=K_TUNE,
        shuffle=True,
        random_state=seed
    )

    folds = list(
        skf.split(
            train_val_df,
            y_tv,
            groups_tv
        )
    )

    for i, (tr_idx, va_idx) in enumerate(
        folds,
        start=1
    ):

        train_cases = set(
            train_val_df
            .iloc[tr_idx]["case_id"]
        )

        val_cases = set(
            train_val_df
            .iloc[va_idx]["case_id"]
        )

        assert train_cases.isdisjoint(
            val_cases
        ), (
            f"Patient leakage detected "
            f"in fold {i}!"
        )

        print(
            f"Seed {seed} | "
            f"Fold {i}: "
            f"Train cases = "
            f"{len(train_cases)}, "
            f"Validation cases = "
            f"{len(val_cases)}, "
            f"Validation images = "
            f"{len(va_idx)}"
        )

    return folds




def cv_fitness(
    learning_rate,
    batch_size,
    weight_decay,
    dropout_rate,
    tune_folds
):

    fold_losses = []

    for fold_no, (
        tr_idx,
        va_idx
    ) in enumerate(
        tune_folds,
        start=1
    ):

        keras.backend.clear_session()



        fold_train = (
            train_val_df
            .iloc[tr_idx]
            .reset_index(drop=True)
        )

        fold_val = (
            train_val_df
            .iloc[va_idx]
            .reset_index(drop=True)
        )

        train_ds = make_dataset(
            fold_train,
            batch_size=int(batch_size),
            training=True
        )

        val_ds = make_dataset(
            fold_val,
            batch_size=int(batch_size),
            training=False
        )



        model, _ = BUILD_FN(
            dropout_rate=float(
                dropout_rate
            )
        )

        compile_model(
            model,
            learning_rate=float(
                learning_rate
            ),
            weight_decay=float(
                weight_decay
            )
        )


        # Training


        history = model.fit(
            train_ds,
            validation_data=val_ds,
            epochs=EPOCHS_PER_CANDIDATE,
            callbacks=[
                keras.callbacks.EarlyStopping(
                    monitor="val_loss",
                    patience=PATIENCE,
                    restore_best_weights=True
                )
            ],
            verbose=0
        )



        img_prob = (
            model
            .predict(
                val_ds,
                verbose=0
            )
            .ravel()
        )



        tmp = pd.DataFrame({

            "case_id":
                fold_val["case_id"].values,

            "y_true":
                (
                    fold_val["class"]
                    == "Stroke"
                ).astype(int).values,

            "y_prob":
                img_prob
        })

        patient_results = (
            tmp
            .groupby("case_id")
            .agg(
                y_true=(
                    "y_true",
                    "first"
                ),
                y_prob=(
                    "y_prob",
                    "mean"
                )
            )
            .reset_index()
        )



        probabilities = np.clip(
            patient_results[
                "y_prob"
            ].values,
            1e-7,
            1 - 1e-7
        )

        loss = log_loss(
            patient_results[
                "y_true"
            ].values,
            probabilities,
            labels=[0, 1]
        )

        fold_losses.append(
            float(loss)
        )

        del model
        keras.backend.clear_session()

    return (
        float(
            np.mean(fold_losses)
        ),
        fold_losses
    )



#  BAYESIAN OPTIMIZATION FOR ONE SEED

def run_bayesian_optimization(
    seed,
    log_path
):



    tune_folds = create_tuning_folds(
        seed
    )



    if (
        os.path.exists(log_path)
        and not RESUME
    ):

        os.remove(log_path)


    records = (
        pd.read_csv(
            log_path
        ).to_dict("records")
        if os.path.exists(log_path)
        else []
    )


    # Bayesian optimizer


    optimizer = Optimizer(
        dimensions=SEARCH_SPACE,
        base_estimator="GP",
        acq_func="EI",
        random_state=seed,
        n_initial_points=10
    )



    for record in records:

        optimizer.tell(
            [
                float(
                    record[
                        "learning_rate"
                    ]
                ),
                int(
                    record[
                        "batch_size"
                    ]
                ),
                float(
                    record[
                        "weight_decay"
                    ]
                ),
                float(
                    record[
                        "dropout_rate"
                    ]
                )
            ],
            float(
                record[
                    "cv_val_loss"
                ]
            )
        )


    # Bayesian optimization loop


    for i in range(
        len(records),
        N_EVALUATIONS
    ):



        candidate = optimizer.ask()

        learning_rate = float(
            candidate[0]
        )

        batch_size = int(
            candidate[1]
        )

        weight_decay = float(
            candidate[2]
        )

        dropout_rate = float(
            candidate[3]
        )



        tf.keras.utils.set_random_seed(
            seed * 1000 + i
        )

        t0 = time.time()



        fit, fold_losses = cv_fitness(
            learning_rate=
                learning_rate,

            batch_size=
                batch_size,

            weight_decay=
                weight_decay,

            dropout_rate=
                dropout_rate,

            tune_folds=
                tune_folds
        )



        optimizer.tell(
            candidate,
            fit
        )



        previous_best = min(
            [
                float(
                    r["cv_val_loss"]
                )
                for r in records
            ],
            default=np.inf
        )

        best_so_far = min(
            fit,
            previous_best
        )

        elapsed = (
            time.time() - t0
        )


        records.append({

            "seed":
                seed,

            "evaluation":
                i + 1,

            "learning_rate":
                learning_rate,

            "batch_size":
                batch_size,

            "weight_decay":
                weight_decay,

            "dropout_rate":
                dropout_rate,

            "cv_val_loss":
                fit,

            **{
                f"fold{k+1}_val_loss":
                    loss
                for k, loss
                in enumerate(
                    fold_losses
                )
            },

            "best_so_far":
                best_so_far,

            "seconds":
                round(
                    elapsed,
                    1
                )
        })



        pd.DataFrame(
            records
        ).to_csv(
            log_path,
            index=False
        )





    log = pd.DataFrame(
        records
    )

    best = log.loc[
        log["cv_val_loss"].idxmin()
    ]

    return best, log



best_by_seed = {}

per_seed = []


for seed in SEEDS:



    best, log = run_bayesian_optimization(
        seed,
        f"{LOG_PREFIX}_seed{seed}_log.csv"
    )

    best_by_seed[
        seed
    ] = best

    fold_cols = [
        f"fold{k+1}_val_loss"
        for k in range(K_TUNE)
    ]

    per_seed.append({

        "seed":
            seed,

        "best_val_loss":
            float(
                best[
                    "cv_val_loss"
                ]
            ),

        "fold_losses":
            [
                float(
                    best[c]
                )
                for c in fold_cols
            ],

        "time_min":
            float(
                log[
                    "seconds"
                ].sum()
                / 60.0
            )
    })




COLS = [
    "Optimization Strategy",
    "Best Validation Loss (Mean ± SD)",
    "95% CI",
    "Mean Optimization Time per Run (min)"
]


def ci95(values):

    v = np.asarray(
        values,
        dtype=float
    )

    n = len(v)

    if n < 2:

        return (
            float(v.mean()),
            np.nan,
            np.nan,
            np.nan
        )

    mean = v.mean()

    sd = v.std(
        ddof=1
    )

    half = (
        stats.t.ppf(
            0.975,
            n - 1
        )
        * sd
        / np.sqrt(n)
    )

    return (
        mean,
        sd,
        mean - half,
        mean + half
    )




vals = [
    r["best_val_loss"]
    for r in per_seed
]

times = [
    r["time_min"]
    for r in per_seed
]


m, sd, lo, hi = ci95(
    vals
)


overall_table = pd.DataFrame(
    [{
        COLS[0]:
            METHOD_NAME,

        COLS[1]:
            f"{m:.4f} ± {sd:.4f}",

        COLS[2]:
            f"{lo:.4f}–{hi:.4f}",

        COLS[3]:
            f"{np.mean(times):,.0f}"
    }],
    columns=COLS
)





seed_table = pd.DataFrame(
    [
        {
            COLS[0]:
                f"{METHOD_NAME} "
                f"(seed {r['seed']})",

            COLS[1]:
                (
                    f"{np.mean(r['fold_losses']):.4f} "
                    f"± "
                    f"{np.std(r['fold_losses'], ddof=1):.4f}"
                ),

            COLS[2]:
                (
                    f"{ci95(r['fold_losses'])[2]:.4f}"
                    "–"
                    f"{ci95(r['fold_losses'])[3]:.4f}"
                ),

            COLS[3]:
                f"{r['time_min']:,.0f}"
        }

        for r in per_seed
    ],
    columns=COLS
)


seed_table.to_csv(
    f"{LOG_PREFIX}_per_seed_table.csv",
    index=False
)


overall_table.to_csv(
    f"table12_{LOG_PREFIX}.csv",
    index=False
)


pd.DataFrame(
    [
        {
            k: v
            for k, v
            in r.items()
            if k != "fold_losses"
        }
        for r in per_seed
    ]
).to_csv(
    f"{LOG_PREFIX}_per_seed_values.csv",
    index=False
)




overall_best = min(
    best_by_seed.values(),
    key=lambda b:
        b["cv_val_loss"]
)


bayesian_best_params = {

    "learning_rate":
        float(
            overall_best[
                "learning_rate"
            ]
        ),

    "batch_size":
        int(
            overall_best[
                "batch_size"
            ]
        ),

    "weight_decay":
        float(
            overall_best[
                "weight_decay"
            ]
        ),

    "dropout_rate":
        float(
            overall_best[
                "dropout_rate"
            ]
        )
}






json.dump(
    {
        "python":
            platform.python_version(),

        "tensorflow":
            tf.__version__,

        "gpus":
            [
                g.name
                for g in
                tf.config.list_physical_devices(
                    "GPU"
                )
            ],

        "SEEDS":
            SEEDS,

        "K_TUNE":
            K_TUNE,

        "N_EVALUATIONS":
            N_EVALUATIONS,

        "EPOCHS_PER_CANDIDATE":
            EPOCHS_PER_CANDIDATE,

        "PATIENCE":
            PATIENCE,

        "SEARCH_SPACE":
            {
                "learning_rate":
                    [1e-5, 1e-3],

                "batch_size":
                    [8, 16, 32],

                "weight_decay":
                    [1e-5, 1e-3],

                "dropout_rate":
                    [0.10, 0.50]
            },

        "fitness":
            "patient-level log loss",

        "dataset":
            "combined train_df + val_df",

        "grouping":
            "case_id",

        "optimizer":
            "Gaussian Process Bayesian Optimization",

        "acquisition_function":
            "Expected Improvement"
    },

    open(
        f"{LOG_PREFIX}_environment.json",
        "w"
    ),

    indent=2
)


json.dump(
    bayesian_best_params,
    open(
        f"{LOG_PREFIX}_best_config.json",
        "w"
    ),
    indent=2
)


vals = np.asarray(
    [r["best_val_loss"] for r in per_seed],
    dtype=float
)

times = np.asarray(
    [r["time_min"] for r in per_seed],
    dtype=float
)

# Mean and SD across the 5 independent seeds
mean_val_loss = vals.mean()
sd_val_loss = vals.std(ddof=1)

# 95% confidence interval across the 5 seeds
n = len(vals)
t_critical = stats.t.ppf(0.975, n - 1)
margin = t_critical * sd_val_loss / np.sqrt(n)

ci_lower = mean_val_loss - margin
ci_upper = mean_val_loss + margin

# Mean optimization time across the 5 seeds
mean_time = times.mean()



print(
    f"{mean_val_loss:.4f} ± {sd_val_loss:.4f}    "
    f"{ci_lower:.4f}–{ci_upper:.4f}    "
    f"{mean_time:,.0f}"
)
```
